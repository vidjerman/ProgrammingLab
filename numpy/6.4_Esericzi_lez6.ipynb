# 5. Analisi di un dataset di video giochi
1. Quanti videogiochi sono stati pubblicati
2. Quali sono i generi più popolari? Mostralo con un bar plot
3. Plotta l'evoluzione del numero di giochi pubblicati nel tempo
4. Prova a riprodurre questo plot 




```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
df.head()
```

```python
len(df)
```

```python
generi=df["Genre"].value_counts()

generi.plot(kind="bar")

plt.title("Numero di videoiochi per genere")
plt.xlabel("Genere")
plt.ylabel("Numero di videogiochi")
plt.xticks(rotation=45)
plt.show()
```

```python
giochi_per_anno=df["Year"].value_counts().sort_index()

giochi_per_anno=df.groupby("Year").size()
giochi_per_anno.plot(kind="line", marker="o")

plt.title("Numero di videoiochi pubblicati nel tempo")
plt.xlabel("Anno")
plt.ylabel("Numero di videogiochi")
plt.grid(True)
```

```python
generi=["Action", "Misc", "Role-Playing", "Shooter", "Sports"]

vendite_genere=df.groupby("Genre")[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]].sum()
vendite_genere = vendite_genere.loc[generi]

vendite_genere.plot(kind="bar", stacked=True, figsize=(10, 5))
plt.title("Vendite per regione per genere")
plt.xlabel("Genre")
plt.ylabel("Milioni di unità")
plt.xticks(rotation=90)
plt.legend()
```
